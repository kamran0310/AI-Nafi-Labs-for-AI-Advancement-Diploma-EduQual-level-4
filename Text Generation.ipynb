```python
import pandas as pd
import numpy as np
from keras.models import Sequential
from keras.layers import LSTM, Dense, Embedding
from keras.utils import get_file
```

```python
#loading shakespare data
path = get_file('shakespeare.txt', origin='https://storage.googleapis.com/download.tensorflow.org/data/shakespeare.txt')
text = open(path).read().lower()
print(text[:500])
```

```python
chars = sorted(list(set(text)))
print(chars)
```

```python
char_to_index = dict((c, i) for i, c in enumerate(chars))
index_to_char = dict((i, c) for i, c in enumerate(chars))
```

```python
print('Vocabulary:', chars)
print('Vocab Size:', len(chars))
print('char_to_index', char_to_index)
print('index_to_char', index_to_char)
```

```python
#prepare the data
max_len = 40
step = 3
sentences = []
next_chars = []

for i in range(0, len(text) - max_len, step):
    sentences.append(text[i: i + max_len])
    next_chars.append(text[i + max_len])
print(sentences[:5])
print(next_chars[:5])
```

```python
# vectoize the data
X = np.zeros((len(sentences), max_len, len(chars)), dtype=np.bool)
y = np.zeros((len(sentences), len(chars)), dtype=np.bool)

for i, sentence in enumerate(sentences):
    for t, char in enumerate(sentence):
        X[i, t, char_to_index[char]] = 1
    y[i, char_to_index[next_chars[i]]] = 1
```

```python
print(len(X[0]))
print(X[0][0])
print(y[0])
```

```python
# Build the LSTM Mode

model = Sequential()
model.add(LSTM(128, input_shape=(max_len, len(chars))))
model.add(Dense(len(chars), activation='softmax'))
model.compile(loss='categorical_crossentropy', optimizer='adam')
model.fit(X, y, epochs = 20, batch_size=128)
```

```python
# funtion to generate text using the trained model
def generate_text(seed_text, length=400):
  generated_text = seed_text.lower()
  for _ in range(length):
    X_pred = np.zeros((1, max_len, len(chars)))
    for t, char in enumerate(seed_text):
      X_pred[0, t, char_to_index[char]] = 1
    pred = model.predict(X_pred, verbose=0)[0]
    next_index = np.argmax(pred)
    next_char = index_to_char[next_index]
    generated_text += next_char
    seed_text = seed_text[1:] + next_char
  return generated_text
```

```python
# generate text starting with the seed.
seed_text = 'to be or not to be'
generated_text = generate_text(seed_text, length=500)
print(generated_text)

```
